```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("sudalairajkumar/covid19-in-india")

print("Path to dataset files:", path)
```

```python
import os
print(os.listdir(path))
```

```python
import pandas as pd
file_path = os.path.join(path, "covid_19_clean_complete.csv")
df=pd.read_csv(file_path)

```

```python
df["Date"] = pd.to_datetime(df["Date"])
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df["Province/State"] = df["Province/State"].fillna("Unknown")
```

```python
df.isnull().sum()
```

```python
df.duplicated()
```

```python
df.head(1)
```

```python
state_data = (df.groupby("Province/State")[["Confirmed","Deaths","Recovered","Active"]].sum().reset_index())
```

```python
state_data.head()
```

```python
import numpy as np
state_data["Death_Rate"] = state_data["Deaths"]/state_data["Confirmed"].replace(0,np.nan)*100
state_data["Recovery_Rate"] = (state_data["Recovered"]/state_data["Confirmed"]).replace(0,np.nan)*100
```

```python
state_data.head()
```

```python
state_data["Recovery_Rate"] = state_data["Recovery_Rate"].round(2)
state_data["Death_Rate"] = state_data["Death_Rate"].round(2)
```

```python
from matplotlib import figure
#Tatal cases by State
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12,5))
sns.barplot(
    data=state_data.sort_values("Confirmed",ascending=False),
    y="Province/State",
    x="Confirmed"
)
plt.title("Total Cases by State")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
