```python
import random

dice = random.randint(1, 6)

print("Dice roll:", dice)

if dice == 4:
    print("4 aaya!")
else:
    print("4 nahi aaya.")

probability = 1 / 6

print("Probability of getting 4:", probability)
print("Probability in percentage:", probability * 100, "%")
```

```python
import random

total_rolls = int(input("Enter the total number of rolls: "))
count_4 = 0

for i in range(total_rolls):
    dice = random.randint(1, 6)

    if dice == 4:
        count_4 += 1

experimental_probability = count_4 / total_rolls

print("Total rolls:", total_rolls)
print("4 came:", count_4, "times")
print("Experimental probability:", experimental_probability)
print("Percentage:", experimental_probability * 100, "%")
```

```python
import pandas as pd
data = {
    "Customer" : range(1,16) ,
    "Visited_before" :["yes","no","yes","no","yes","no","no","no","yes","no","yes","no","yes","yes","yes"] ,
    "Added_to_card" : ["yes","no","no","yes","no","no","yes","no","yes","no","yes","no","yes","no","yes"] ,
    "Purchased" : ["yes","no","yes","no","yes","no","yes","no","no","no","yes","no","yes","no","yes"]

}
df = pd.DataFrame(data)
print(df)
```

```python
purchased_probability = (
    df["Purchased"] == "yes"
).mean()

print(purchased_probability)
```

```python
# what is the probability of purchase given that he cutomer added something to cart?
purchased_given_added_to_card = (
    df[df["Added_to_card"] == "yes"]["Purchased"] == "yes"
).mean()

#print(purchased_given_added_to_card)
print(purchased_given_added_to_card * 100, "%")
```

```python
from scipy.stats import norm
import numpy as np
import matplotlib.pyplot as plt
mean = 78
std = 10
x= np.linspace(30,110,500)
y=norm.pdf(x, mean, std)
plt.plot(x,y)

plt.xlabel("Exam Scores")
plt.ylabel("Probability Density")
plt.title("Normal Distribution of Exam Scores")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

n=10
p = 0.30
x= np.arange(0,n+1)
probabilites = binom.pmf(x,n,p)
plt.bar(x,probabilites)
plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.title("Binomial Distribution : 10 customers, purchase probability =")
plt.show()
```
